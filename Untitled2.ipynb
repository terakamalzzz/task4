```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.plot(days, temperature)

plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
plt.plot(
    days,
    temperature,
    marker="o",
    linestyle="--"
)

plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
plt.plot(days, temperature, marker="o", linestyle="--")

plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.grid()

plt.show()
```

```python
plt.plot(days, temperature, marker="o", linestyle="--")

for i, val in enumerate(temperature):
    plt.text(i, val, str(val))

plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    days,
    temperature,
    marker="o",
    linestyle="--"
)

for i, val in enumerate(temperature):
    plt.text(i, val, str(val))

plt.title("Weekly Tewmprature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.grid()

plt.show()
```

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]

students = [35, 28, 22, 18, 15]
```

```python
plt.figure(figsize=(10, 5))

plt.barh(languages, students)

plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.barh(
    languages,
    students,
    color=["blue", "green", "red", "orange", "purple"]
)

plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.barh(
    languages,
    students,
    color=["blue", "green", "red", "orange", "purple"]
)

for i, val in enumerate(students):
    plt.text(val, i, str(val))

plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")

plt.show()
```

```python
colors = ["red", "blue", "blue", "blue", "blue"]

plt.figure(figsize=(10, 5))

plt.barh(languages, students, color=colors)

for i, val in enumerate(students):
    plt.text(val, i, str(val))

plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]

fig = plt.figure(figsize=(12, 6))
axes1 = fig.add_axes([0.08, 0.12, 0.55, 0.78])

axes1.scatter(study_hours, scores)

plt.show()
```

```python
fig = plt.figure(figsize=(12, 6))
axes1 = fig.add_axes([0.08, 0.12, 0.55, 0.78])

scatter = axes1.scatter(
    study_hours,
    scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=150
)

plt.show()
```

```python
fig = plt.figure(figsize=(12, 6))
axes1 = fig.add_axes([0.08, 0.12, 0.55, 0.78])

scatter = axes1.scatter(
    study_hours, scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=150
)

cbar = fig.colorbar(scatter, ax=axes1)
cbar.set_label("Age")

plt.show()
```

```python
fig = plt.figure(figsize=(12, 6))
axes1 = fig.add_axes([0.08, 0.12, 0.55, 0.78])

scatter = axes1.scatter(
    study_hours, scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=150
)

cbar = fig.colorbar(scatter, ax=axes1)
cbar.set_label("Age")

axes1.set_title("Students' Exam Results")
axes1.set_xlabel("Study Hours")
axes1.set_ylabel("Exam Score")

plt.show()
```

```python
max_score = max(scores)

max_index = scores.index(max_score)

print("Highest Score:", max_score)
print("Study Hours:", study_hours[max_index])
print("Age:", ages[max_index])
```
